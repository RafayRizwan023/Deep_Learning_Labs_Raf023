```python
# Step 1 — Importing Required Libraries

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import classification_report, accuracy_score
import matplotlib.pyplot as plt
```

```python

# Load the Iris dataset
iris = load_iris()

X = iris.data       # Features
y = iris.target     # Labels

# Split the data into training and testing sets
# 70% training, 30% testing
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=42
)

```

```python
# Standardize the features to have mean = 0
# and standard deviation = 1
scaler = StandardScaler()

# Fit on training data and transform
X_train_scaled = scaler.fit_transform(X_train)

# Transform test data using the same scaler
X_test_scaled = scaler.transform(X_test)
```

```python
# Create an MLP classifier with:
# - Two hidden layers
# - 10 neurons in each hidden layer
# - Maximum 1000 iterations
# - Random state = 42
# - Learning rate = 0.001

mlp = MLPClassifier(
    hidden_layer_sizes=(10, 10),
    max_iter=1000,
    random_state=42,
    learning_rate_init=0.001
)

# Train the MLP classifier
mlp.fit(X_train_scaled, y_train)


```

```python
# Predict the test set
y_pred = mlp.predict(X_test_scaled)

# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)

print(f"Accuracy: {accuracy:.2f}")

# Print detailed classification report
print("Classification Report:\n", classification_report(y_test, y_pred))

```

```python

print("MLP Structure:")
print(f"Number of layers: {mlp.n_layers_}")
print(f"Number of outputs: {mlp.n_outputs_}")
print(f"Activation function: {mlp.activation}")
print(f"Output activation function: {mlp.out_activation_}")
print(f"Number of epochs: {mlp.n_iter_}")
```

```python
# Plot the loss curve
plt.figure(figsize=(8, 6))

plt.plot(
    mlp.loss_curve_,
    label='Training Loss'
)

plt.title('MLP Classifier Learning Curve')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()
plt.grid()

plt.show()
```

We observe that changing the learning rate affects the number of epochs required for the model to converge and, consequently, the behavior of the loss function. In general, increasing the learning rate allows the model to update its weights more rapidly, which can lead to faster convergence. However, if the learning rate is too high, the model may overshoot the minimum loss and require more epochs to reach a stable or optimal loss value.
